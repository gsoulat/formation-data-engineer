# Mini-brief ML #09 — PCA (réduction de dimension)

> Objectif : **variance expliquée cumulée** pour choisir n_components, **projection 2D**, PCA en amont d'un modèle.

Dataset : `load_digits` (64 dimensions).

```python
import numpy as np, matplotlib.pyplot as plt
from sklearn.datasets import load_digits
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

X, y = load_digits(return_X_y=True)
Xs = StandardScaler().fit_transform(X)
X.shape
```

## 1. Le geste clé — combien de composantes garder ?

```python
pca = PCA().fit(Xs)
cum = np.cumsum(pca.explained_variance_ratio_)
plt.plot(cum, marker='.'); plt.axhline(0.9, color='r', ls='--'); plt.xlabel('n composantes'); plt.ylabel('variance cumulée'); plt.title('Variance expliquée'); plt.show()
print('composantes pour 90% :', int(np.argmax(cum>=0.9))+1)
```

## 2. Projection 2D lisible

```python
P = PCA(n_components=2).fit_transform(Xs)
plt.figure(figsize=(7,6)); sc=plt.scatter(P[:,0], P[:,1], c=y, cmap='tab10', s=12, alpha=.7); plt.colorbar(sc); plt.title('digits projeté en 2D'); plt.show()
```
