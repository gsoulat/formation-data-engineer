# Mini-brief ML #03 — Régression logistique

> Objectif : entraîner, **interpréter (odds ratio)**, évaluer (ROC-AUC), et **déplacer le seuil** selon le coût métier.

Dataset : `load_breast_cancer` (binaire).

```python
import numpy as np, matplotlib.pyplot as plt
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import make_pipeline
from sklearn.metrics import roc_auc_score, confusion_matrix, classification_report

X, y = load_breast_cancer(return_X_y=True, as_frame=True)
Xtr, Xte, ytr, yte = train_test_split(X, y, test_size=.25, stratify=y, random_state=42)
```

## 1. Baseline (pipeline + normalisation)

```python
clf = make_pipeline(StandardScaler(), LogisticRegression(max_iter=5000)).fit(Xtr, ytr)
proba = clf.predict_proba(Xte)[:,1]
print('ROC-AUC :', round(roc_auc_score(yte, proba),3))
```

## 2. Le geste clé — déplacer le seuil de décision
Par défaut on classe à p≥0.5. Si rater un cas positif coûte cher, on **baisse le seuil** pour gagner du rappel.

```python
for s in [0.5, 0.3, 0.2]:
    pred = (proba >= s).astype(int)
    tn, fp, fn, tp = confusion_matrix(yte, pred).ravel()
    print(f'seuil {s}: faux négatifs={fn}, faux positifs={fp}')
```
