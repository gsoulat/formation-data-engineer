# Mini-brief ML #07 — SVM (machine à vecteurs de support)

> Objectif : **normalisation obligatoire**, effet de **kernel / C / gamma**, frontière de décision.

Dataset : `load_breast_cancer`.

```python
import numpy as np, matplotlib.pyplot as plt
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.pipeline import make_pipeline

X, y = load_breast_cancer(return_X_y=True, as_frame=True)
Xtr, Xte, ytr, yte = train_test_split(X, y, test_size=.25, stratify=y, random_state=42)
```

## 1. SVM linéaire vs RBF (toujours normalisé)

```python
for k in ['linear','rbf']:
    svc = make_pipeline(StandardScaler(), SVC(kernel=k)).fit(Xtr, ytr)
    print(f'kernel={k}: test={svc.score(Xte,yte):.3f}')
```

## 2. Le geste clé — C et gamma (kernel RBF)
`C` = tolérance aux erreurs (grand C = marge étroite) ; `gamma` = portée d'un point (grand gamma = surapprentissage local).

```python
for C in [0.1, 1, 10]:
    for g in ['scale', 0.01, 1]:
        s = make_pipeline(StandardScaler(), SVC(kernel='rbf', C=C, gamma=g)).fit(Xtr, ytr)
        print(f'C={C}, gamma={g}: test={s.score(Xte,yte):.3f}')
```
