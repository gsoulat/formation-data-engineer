# Mini-brief ML #10 — Naive Bayes

> Objectif : entraîner un modèle **très rapide**, comprendre l'**hypothèse d'indépendance**, comparer à la régression logistique.

Dataset : `load_breast_cancer` (GaussianNB). Pour du texte, on utiliserait `MultinomialNB` + TF-IDF.

```python
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.metrics import accuracy_score
import time

X, y = load_breast_cancer(return_X_y=True, as_frame=True)
Xtr, Xte, ytr, yte = train_test_split(X, y, test_size=.25, stratify=y, random_state=42)
```

## 1. Rapidité — l'atout de Naive Bayes

```python
t=time.time(); nb=GaussianNB().fit(Xtr, ytr); tnb=time.time()-t
print(f'GaussianNB : accuracy={nb.score(Xte,yte):.3f}  temps={tnb*1000:.1f} ms')
```

## 2. Le geste clé — comparer à la régression logistique

```python
t=time.time(); lr=make_pipeline(StandardScaler(), LogisticRegression(max_iter=5000)).fit(Xtr, ytr); tlr=time.time()-t
print(f'LogReg     : accuracy={lr.score(Xte,yte):.3f}  temps={tlr*1000:.1f} ms')
```
