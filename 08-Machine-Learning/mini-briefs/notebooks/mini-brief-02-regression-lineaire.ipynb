# Mini-brief ML #02 — Régression linéaire

> Objectif : **interpréter les coefficients**, lire R², analyser les résidus, comparer Ridge/Lasso.

Dataset : `load_diabetes` (régression réelle).

```python
import numpy as np, matplotlib.pyplot as plt
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.metrics import r2_score, mean_absolute_error

X, y = load_diabetes(return_X_y=True, as_frame=True)
Xtr, Xte, ytr, yte = train_test_split(X, y, test_size=.25, random_state=42)
```

## 1. Baseline + interprétation des coefficients
Un coefficient positif = quand la variable augmente, la cible augmente (toutes choses égales par ailleurs).

```python
lin = LinearRegression().fit(Xtr, ytr)
pred = lin.predict(Xte)
print('R2 :', round(r2_score(yte, pred),3), '| MAE :', round(mean_absolute_error(yte, pred),1))
import pandas as pd
print(pd.Series(lin.coef_, index=X.columns).sort_values())
```

## 2. Résidus — le diagnostic honnête
Les résidus doivent être ~centrés sur 0 et sans structure. Une forme en 'entonnoir' = hétéroscédasticité.

```python
resid = yte - pred
plt.scatter(pred, resid, alpha=.6); plt.axhline(0, color='r'); plt.xlabel('prédit'); plt.ylabel('résidu'); plt.title('Résidus'); plt.show()
```
