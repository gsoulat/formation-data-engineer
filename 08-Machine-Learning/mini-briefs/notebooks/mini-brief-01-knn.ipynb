# Mini-brief ML #01 — KNN (k plus proches voisins)

> Version notebook exécutable. Objectif : **utiliser** KNN — normaliser, choisir k, évaluer.

Dataset : `load_wine` (3 classes, à normaliser).

```python
import numpy as np, matplotlib.pyplot as plt
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.pipeline import make_pipeline

X, y = load_wine(return_X_y=True, as_frame=True)
Xtr, Xte, ytr, yte = train_test_split(X, y, test_size=.25, stratify=y, random_state=42)
X.shape
```

## 1. Baseline — et pourquoi la normalisation est OBLIGATOIRE
KNN mesure des distances : une variable à grande échelle écrase les autres. On met le `StandardScaler` **dans un pipeline** (pas de fuite de données).

```python
knn = make_pipeline(StandardScaler(), KNeighborsClassifier(n_neighbors=5))
knn.fit(Xtr, ytr)
print('accuracy test :', round(knn.score(Xte, yte), 3))
```

## 2. Le geste clé — choisir k par validation croisée

```python
ks = range(1, 21)
scores = [cross_val_score(make_pipeline(StandardScaler(), KNeighborsClassifier(n_neighbors=k)), Xtr, ytr, cv=5).mean() for k in ks]
plt.plot(list(ks), scores, marker='o'); plt.xlabel('k'); plt.ylabel('accuracy CV'); plt.title('Choix de k'); plt.show()
print('meilleur k :', list(ks)[int(np.argmax(scores))])
```
