# Mini-brief ML #04 — Arbre de décision

> Objectif : entraîner, **visualiser l'arbre**, montrer l'**overfitting** quand il est trop profond, lire les importances.

Dataset : `load_iris`.

```python
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree

X, y = load_iris(return_X_y=True, as_frame=True)
Xtr, Xte, ytr, yte = train_test_split(X, y, test_size=.3, stratify=y, random_state=42)
```

## 1. Un arbre lisible (profondeur limitée)

```python
tree = DecisionTreeClassifier(max_depth=3, random_state=42).fit(Xtr, ytr)
print('accuracy test :', round(tree.score(Xte, yte),3))
plt.figure(figsize=(11,6)); plot_tree(tree, feature_names=X.columns, class_names=True, filled=True); plt.show()
```

## 2. Le geste clé — voir l'overfitting
Un arbre sans limite mémorise le train (accuracy train ~1.0) mais généralise mal.

```python
for d in [1,2,3,5,None]:
    t = DecisionTreeClassifier(max_depth=d, random_state=42).fit(Xtr, ytr)
    print(f'max_depth={d}: train={t.score(Xtr,ytr):.3f}  test={t.score(Xte,yte):.3f}')
```
