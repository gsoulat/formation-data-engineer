# Mini-brief ML #08 — K-Means (clustering non supervisé)

> Objectif : **choisir k** (coude + silhouette), normaliser, **interpréter les clusters**.

Dataset : `load_iris` (on ignore les étiquettes — non supervisé).

```python
import numpy as np, matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

X, _ = load_iris(return_X_y=True, as_frame=True)
Xs = StandardScaler().fit_transform(X)
```

## 1. Le geste clé — la méthode du coude

```python
inertias=[]; sils=[]
ks=range(2,9)
for k in ks:
    km = KMeans(n_clusters=k, n_init=10, random_state=42).fit(Xs)
    inertias.append(km.inertia_); sils.append(silhouette_score(Xs, km.labels_))
fig,ax=plt.subplots(1,2,figsize=(11,4))
ax[0].plot(list(ks), inertias, marker='o'); ax[0].set_title('Coude (inertie)')
ax[1].plot(list(ks), sils, marker='o'); ax[1].set_title('Silhouette'); plt.show()
```

## 2. Interpréter les clusters
Une fois k choisi, on regarde le profil moyen de chaque cluster pour lui donner un sens métier.

```python
import pandas as pd
km = KMeans(n_clusters=3, n_init=10, random_state=42).fit(Xs)
print(pd.DataFrame(X).groupby(km.labels_).mean().round(2))
```
