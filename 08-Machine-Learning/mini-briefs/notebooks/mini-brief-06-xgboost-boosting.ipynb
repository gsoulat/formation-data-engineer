# Mini-brief ML #06 — Gradient Boosting (esprit XGBoost)

> Objectif : **learning rate** vs **n_estimators**, **early stopping**, courbe de validation.

Dataset : `load_breast_cancer`. On utilise `HistGradientBoostingClassifier` (intégré, early stopping natif). L'API XGBoost est très proche.

```python
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.ensemble import HistGradientBoostingClassifier

X, y = load_breast_cancer(return_X_y=True, as_frame=True)
Xtr, Xte, ytr, yte = train_test_split(X, y, test_size=.25, stratify=y, random_state=42)
```

## 1. Early stopping — laisser le modèle s'arrêter tout seul

```python
gb = HistGradientBoostingClassifier(learning_rate=0.1, max_iter=500, early_stopping=True, validation_fraction=0.2, random_state=42).fit(Xtr, ytr)
print('itérations réellement utilisées :', gb.n_iter_)
print('accuracy test :', round(gb.score(Xte, yte),3))
```

## 2. Le geste clé — le compromis learning_rate / n_iter
Un petit learning_rate apprend plus finement mais demande plus d'itérations.

```python
for lr in [0.3, 0.1, 0.03]:
    m = HistGradientBoostingClassifier(learning_rate=lr, max_iter=500, early_stopping=True, random_state=42).fit(Xtr, ytr)
    print(f'lr={lr}: n_iter={m.n_iter_:>3}  test={m.score(Xte,yte):.3f}')
```
