## 1. Qu'est-ce que Python ?

### Histoire et Créateur

Python est un langage de programmation créé par **Guido van Rossum** en 1991. Le nom "Python" vient de la troupe de comédiens britanniques "Monty Python", dont Guido était fan.

**Chronologie importante :**
- **1991** : Python 0.9.0 - Première version publique
- **2000** : Python 2.0 - Introduction des list comprehensions
- **2008** : Python 3.0 - Refonte majeure (non rétrocompatible)
- **2020** : Fin du support de Python 2
- **Aujourd'hui** : Python 3.x est la seule version maintenue

### Philosophie de Python

Python a été conçu avec l'objectif d'être **simple, lisible et explicite**. La philosophie est résumée dans le "Zen de Python" (que nous verrons plus tard).

**Principes clés :**
- La lisibilité compte
- Simple est mieux que complexe
- Il devrait y avoir une seule façon évidente de faire les choses

## 4. Le REPL (Read-Eval-Print Loop)

Le **REPL** est un mode interactif qui permet de tester du code Python ligne par ligne.

### Comment y accéder ?

Dans un terminal, tapez simplement :
```bash

```

Vous verrez quelque chose comme :
```

[GCC 11.4.0] on linux
Type "help", "copyright", "credits" or "license" for more information.

```

Le symbole `>>>` indique que Python attend une commande.

````python
# Le notebook Jupyter fonctionne comme un REPL amélioré
# Exemple : utiliser Python comme calculatrice
2 + 2
````

````python
# La fonction help() donne de la documentation
help(print)
````

**Quitter le REPL :** Tapez `exit()` ou `quit()`, ou utilisez le raccourci `Ctrl+D` (Linux/Mac) ou `Ctrl+Z` puis Entrée (Windows).

## 5. Premier Script Python

### Créer un fichier .py

Un script Python est un fichier texte avec l'extension `.py`.

**Exemple : hello.py**
```python
# Mon premier script Python
print("Bonjour, monde !")
print("Python est génial !")
```

### Exécuter un script

Dans le terminal, naviguez vers le dossier contenant votre script et tapez :
```bash
python hello.py
```

### Différence REPL vs Script

- **REPL** : Affiche automatiquement le résultat de la dernière expression
- **Script** : Nécessite `print()` pour afficher quelque chose

````python
# Dans le REPL (ou un notebook), ceci affiche 4
2 + 2
````

````python
# Dans un script, ceci n'affiche rien (le résultat est perdu)
2 + 2

# Il faut utiliser print() pour voir le résultat
print(2 + 2)
````

## 7. Python 2 vs Python 3

### Pourquoi Python 3 uniquement ?

Python 2.7 a reçu son dernier correctif le **1er janvier 2020**. Il n'est plus maintenu.

**Principales différences :**

| Fonctionnalité | Python 2 | Python 3 |
|----------------|----------|----------|
| print | `print "hello"` | `print("hello")` |
| Division | `5 / 2 = 2` | `5 / 2 = 2.5` |
| Unicode | ASCII par défaut | UTF-8 par défaut |
| range() | Retourne une liste | Retourne un itérateur |
| Support | ❌ Terminé | ✅ Actif |

**Conclusion :** N'utilisez que Python 3 pour tous les nouveaux projets.

````python
# Python 3 : print est une fonction
print("Hello", "World", sep=" - ")

# Python 2 : ceci produirait une erreur car print était un statement
````

````python
# Python 3 : division vraie par défaut
print(5 / 2)   # 2.5
print(5 // 2)  # 2 (division entière)

# Python 2 : 5 / 2 donnait 2
````

### Principes clés à retenir
